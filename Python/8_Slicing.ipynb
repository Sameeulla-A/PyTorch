* Slicing, is a procedure on Lists and tuples and numpy arrays of matrices that allows you to access multiple elements at the same time.
* let's learn another data type and function called range.
* range is reserve key word and a special function.

```python
a = range(5,11)
a
```

* range is a range, exactly as we put it. but this is it's own type.
* This is a seperate special type of data in python.
* range is used in python all the time. when creating for loops.
* range is going to create a set of integers that go from the first number up to but excluding the final number. means we get the numbers from 5 to 10.

```python
type(a)
```

```python
a = list(range(5,11))  #converting range into list.
a
```

* we got the list from 5 to 10, it always an exclusive upper bound, so it doesn't actually gets us up to 10.
* if you want to get the number six

```python
a[1]
```

```python
a[2]
```

* index gave 7.
* slicing works y specifing the lower bound : upper bound.

```python
a[0:2]
```

* here python interprects the upper bound of slicing as exclusive, so this return 0 to 2.
* indexing starts from 0 1 2..
* slicing excludes the upper bound.
* we can even the upper bound.
* so the zero is inplecit

```python
a[:4]
```

* even indecies in the list.

```python
a[::2] # we want all of the elements, but skipping by 2
```

```python
a[1::2]  # here we start from zero, and skipping zero.
```

```python
a[1::4]
```

```python
a[1::5] # we only get six, because index starts at 1 and then we count five. there's nothing else here. so python just stops here.
```

* how do we interprect ::?
* start:stop:skip
* so the first is the index and stop is the index, and this the number of indices that we want to skip by

* how to reverse a list?
* ::-1   
* list comes in forward, instead of going forward, we are skipping backwards.

```python
a[::-1]
```
