* An ndarray (N-dimensional array) is the core data structure of the NumPy library in Python. It is a fast, flexible, and multidimensional container designed to hold homogeneous data.

```python
# convert the list into ndarray
np_nums = np.array(numbers)
print(type(numbers))
print(np_nums)
print(type(np_nums))
```

```python
print(np_nums.mean())
print(np_nums.max())
print(np_nums.min())

# or

print(np.mean(np_nums))
print(np.max(np_nums))
print(np.min(np_nums))
```

```python
# In-Place Operations
# Out-of-Place Operations / not in-place operations
n = [ 4,3,5,2,6,1,7 ]  # unsorted list
print(n)
n.sort()
print(n)
```

* when we apply sort on a list, then it's going to actually change the variable, doesn't just print out the sorted version or return a sorted version as different variable.
* it actually fundamentally changes the variable.
* similar process works with methods in numpy

```python
m = [ 9,5,8,6,7,2,4,3,1 ]
nn = np.array(m) # does the same thing happens with numpy array as above.
print(nn)
nn.sort()
print(nn)
```

```python
# now let's call the numpy function
p = [ 9,5,8,6,7,2,4,3,1 ]
no = np.array(p) # does the same thing happens with numpy array as above.
print(p)
np.sort(no)
print(no)
```

* it doesn't actually change ,  beacause it's an inplace operation

```python
# let's try out place operation
p = [ 9,5,8,6,7,2,4,3,1 ]
no = np.array(p) # does the same thing happens with numpy array as above.
print(p)
op = np.sort(no)
print(op)
```

* **Note:**  certain operations in python operate in-place, meaning variable is changing. and certain operation are not in-pace meaning that they don't change the variable on it's own or you can overwrite the existing variable.
