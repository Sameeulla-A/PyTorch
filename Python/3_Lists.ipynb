* List is more flexible than an individual number or an individual string.
* list encapsulated with [square brackets], different brackets for variables or different operations in python.

* list contains multiple items and seperated by a comma.   
*  [1,2,3,4,5]




```python
alist = [1,2,3,4,5]  #numbers
strlist = ['hi','my','name','is','bob'] #strings
alphnum = [1,'h',3,4,'ff','hhf'] # mix list
print(type(alist))
print(type(strlist))
print(type(alphnum))
```

python doesn't care that the different individual elements have different types, have different classes, it's totally fine, everything goes into a list, as long as seperated by commas.
* spaces, ; any other special characters are not allowed, it througs error.
* you can have a list inside the list but, should be seperated by a comma.

```python
lils = [1,2,3,'hhj',
        ['hhdj','tgetd','qwer'],
        [.5,4/5,],3,4,'j']

print(lils)
print(type(lils))
```

* we can organize the list in any order, but it should be seperated by comma and must be in [].

**Method.**
* method is function that acts on particular class of variables.
* print is a function, used to print the output.
* method is similar to a function, but the syntax is slightly different. at the end we write our variable name and then at the end we arite a dot, so this is a period.
* append is used to add the values to the list.

```python
print(alist)
print(len(alist))
type(alist)
```

```python
print(alist)
alist.append(-100)
print(alist)
```

```python
#let's run the above list again.
print(alist)
alist.append(-100)
print(alist)
```
