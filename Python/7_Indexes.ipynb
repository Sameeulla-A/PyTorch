* Indexing: What allows us to access particular elements of a list or a tuple or any data type that is an ordered list of items, not just lists.
* Indexing means to call a specific index or element of an ordered collection of Information.

```python
alist = [4,6,2,-6,'qwerty',10.01]
'''
lets access -6 from the list. for the human eye -6 is 4th element.
even in some programming languages, math oriented Programming like MATLAB & Julia.
also call this the 4th element of the list. but in Python start from 0 (ZERO).

'''

alist[3]
```

image.png

* in python first index is ZERO(0)
* square brackets are generally used for indexing, regardless of the data type.

```python
btup = (5,3,-5,'hgjt',5.45)  # a tuple

alist[3]
btup[3]
```

* indices have to be integers.
* it doesn't make sense to talk about 4.5th element of a list.

```python
alist[4.5]
```

```python
# float will give the same error.

alist[4.0]
```

```python
#convert the float into int then it will work.
idx = 4.0
alist[int(idx)]
```

```python
'''
incase we don't know the length of list.
we can use len()
'''

alist[len(alist)]
```

```python
len(alist)
```

* the index 6 actually corresponds to the seventh element in this list, but there's no 7th element in the list.

```python
alist[len(alist)-1]
```

* simply you can use -1, means negative indexing.

```python
alist[-1] # last element in the list
```

```python
alist[-2] # second last element in the list
```

* postive indexing starts from 0 from left side.
* negative indexing starts from -1 from right side.


* We can use varibles for indexing.
* what if we want multiple index range.?

```python
idx = [3,5]
alist[idx]
# the above is legal behaviour in MATLAB.
# but python doesn't allow this.
```

```python
idx = [3,5]
alist[int(idx)]
```

```python
idx = (3,5)  #tuples
alist[idx]
```

listlist = [4, 'hi', [5,4,3],'yo',{'squirrel':'cute', 'penguin':'yummy'}]

* get the attribute of penguin from this list.

```python
listlist = [4, 'hi', [5,4,3],'yo',{'squirrel':'cute', 'penguin':'yummy'}]
listlist
```

```python
tmp = listlist[4]
tmp
```

```python
tmp['penguin']
```

```python
listlist[4]['penguin']
```

* With a dictionary, you generally don't use a numeric index. You access a value using its key.

* what if we have nested disctionary?


```python
nstdct = [4, 'hi', [5,4,3],'yo',{'squirrel':'cute', 'penguin':'yummy', 'panda': {'color':'white', 'size':'chubby'}}]
nstdct
```

same way let's extract panda's size

```python
nst = nstdct[4]
nst
```

```python
sz = nst['panda']['size']
sz
```

```python
nst = nstdct[4]['panda']['size']
nst
```
