* function gives output.
* and the way that the inputs and output work is that the inputs go after the function call inside parentheses and the output of the function or result of the alogorith or computation that the function is implementing goes to the left with an equal sign.
* so output equals function name, input.
* and we can print the sum.

```python
output = sum(var)
print('The sum is ' + str(output))  #python doesn't know what to do with string and int so we convert.
```

* sum function can multiple input and provide multiple outputs.
* some functions provide multiple outputs, which can also be seperated by ','

```python
output = sum(var, axis=1) # we can provide multiple inputs into functions by separating them with ','
```
