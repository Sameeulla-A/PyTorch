* Difference between the List and Tuple:
* a list is called as mutable, means you can change the elements once you defined.
* a tuple is immutable, means you cannot change the elements once you defined.

the below list and tuple have same elements and in same order but they have different data types.

```python
atuple = (1,'asdf',4.4)
alist = [1,'asdf',4.4]
```

```python
print(alist)
alist[0] = 13
print(alist)
# here we have updated the list using index.
```

```python
print(atuple)
atuple[0] = 13
print(atuple)
# here we have updated the tuple using index.
```
