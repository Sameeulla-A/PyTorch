* What is a Dictionary?
* The idea of a dictionary is that we ahve key value pairs.
* whatever we store in dictionary has a key and has a value, you cannot have a key with no values and you cannot have values floating around without any keys.
* How does this work?


```python
D = dict()
D
```

* () for tuples
* {} for dictionaries
