* Now these are not the only way to create Booleans. we can also create Boolean variables using comparisons.
* we are making assignments with one equal(=) sign.
* When you write two equal (==) signs, we're actually asking a question and the question is, Is the Left Hand Side Equal to Right Hand Side?
* So the answer to the question is True or False.

```python
5 == 5
```

```python
5 ==3
```

```python
4 == 4.0
```

* for human eyes looks like they are the same number, but these are different data types.
* 4 is an int and the 4.0 is a float.
* Python returns True, it recognizes in this case that although these are different data types, they are different variables types, different classes, numerically, they are still the same thing.

```python
test = 6 == 6
test
```
